```python
import numpy as np

# Scores in [Math, Science, English]
S1 = np.array([80, 75, 85])
S2 = np.array([78, 80, 82])
S3 = np.array([60, 65, 70])

print("Student 1:", S1)
print("Student 2:", S2)
print("Student 3:", S3)
```

```python
print("L2 Norm of S1:", np.linalg.norm(S1))
print("L2 Norm of S2:", np.linalg.norm(S2))
print("L2 Norm of S3:", np.linalg.norm(S3))
```

```python
sim12 = np.dot(S1, S2) / (np.linalg.norm(S1) * np.linalg.norm(S2))
sim13 = np.dot(S1, S3) / (np.linalg.norm(S1) * np.linalg.norm(S3))
sim23 = np.dot(S2, S3) / (np.linalg.norm(S2) * np.linalg.norm(S3))

print("S1-S2:", sim12)
print("S1-S3:", sim13)
print("S2-S3:", sim23)
```

```python
import numpy as np

# x = Pencil, y = Pen, z = Eraser
A = np.array([
    [2, 3, 1],
    [1, 2, 3],
    [3, 1, 2]
])

B = np.array([145, 125, 160])

price = np.linalg.solve(A, B)

print("Pencil =", price[0])
print("Pen =", price[1])
print("Eraser =", price[2])
```
